```python
# Encodes a list of texts into vector embeddings in batches of 50 at a time rather than passing the entire dataset at once. 
# It uses tqdm to display a live progress bar showing elapsed time and progress.
from tqdm.auto import tqdm

batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i:i + batch_size]
    batch_vectors = model.encode(batch)
    vectors.extend(batch_vectors)

len(vectors)
```

```python
from tqdm.auto import tqdm

from ingest import load_currency_data
from sentence_transformers import SentenceTransformer


model = SentenceTransformer('all-MiniLM-L6-v2')

documents = load_currency_data()
texts = []

for doc in documents:
    countries = doc["countries"]

    if isinstance(countries, list):
        countries = ", ".join(countries)

    text = (
        f"Currency code: {doc['code']}. "
        f"Currency name: {doc['name']}. "
        f"Symbol: {doc['symbol']}. "
        f"Decimal digits: {doc['decimal_digits']}. "
        f"Countries: {countries}."
    )

    texts.append(text)

batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i:i + batch_size]
    batch_vectors = model.encode(batch)
    vectors.extend(batch_vectors)
```

```python
import psycopg

conn = psycopg.connect(
    "postgresql://postgres:mysecretpassword@localhost:5431/faq",
    connect_timeout=10
)

conn.execute("CREATE EXTENSION IF NOT EXISTS vector")
```

```python
conn.execute("""
    DROP TABLE IF EXISTS documents
""")

conn.execute("""
    CREATE TABLE IF NOT EXISTS currencies (
    id SERIAL PRIMARY KEY,
    code TEXT,
    name TEXT,
    symbol TEXT,
    decimal_digits INTEGER,
    countries TEXT[],
    embedding VECTOR(384)
)
""")
```

```python
def vec_to_str(vector):
    return '[' + ','.join(str(x) for x in vector) + ']'

for doc, vec in tqdm(zip(documents, vectors), total=len(documents)):
    conn.execute(
        """
        INSERT INTO currencies
            (code, name, symbol, decimal_digits, countries, embedding)
        VALUES (%s, %s, %s, %s, %s, %s::vector)
        """,
        (
            doc["code"],
            doc["name"],
            doc["symbol"],
            doc["decimal_digits"],
            doc["countries"],
            vec_to_str(vec)
        )
    )

conn.commit()
```

```python
query = 'I just discovered the course. Can I still join it?'
query_vector = model.encode(query)
query_str = vec_to_str(query_vector)
```

```python
results = conn.execute(
    """
    SELECT code, name, symbol, decimal_digits, countries,
           1 - (embedding <=> %s::vector) AS similarity
    FROM currencies
    ORDER BY embedding <=> %s::vector
    LIMIT 5
    """,
    (query_str, query_str)
).fetchall()

for row in results:
    print(
        f"Currency: {row[1]} ({row[0]})\n"
        f"Symbol: {row[2]}\n"
        f"Decimal digits: {row[3]}\n"
        f"Countries: {', '.join(row[4])}\n"
        f"Similarity: {row[5]:.4f}\n"
    )
```

```python
results = conn.execute(
    """
    SELECT code, name, symbol, decimal_digits, countries,
           1 - (embedding <=> %s::vector) AS similarity
    FROM currencies
    WHERE code = %s
    ORDER BY embedding <=> %s::vector
    LIMIT 5
    """,
    (query_str, 'JPY', query_str)
).fetchall()

for row in results:
    print(
        f"[{row[0]}] {row[1]} "
        f"(symbol: {row[2]}, similarity: {row[5]:.4f})"
    )
```

```python
def pgvector_search(query, num_results=5):
    query_vector = model.encode(query)
    query_str = vec_to_str(query_vector)

    rows = conn.execute(
        """
        SELECT code, name, symbol, decimal_digits, countries
        FROM currencies
        ORDER BY embedding <=> %s::vector
        LIMIT %s
        """,
        (query_str, num_results)
    ).fetchall()

    return [
        {
            'code': r[0],
            'name': r[1],
            'symbol': r[2],
            'decimal_digits': r[3],
            'countries': r[4]
        }
        for r in rows
    ]
```

```python
from rag_helper import RAGBase


class RAGPgVector(RAGBase):

    def __init__(self, embedder, conn, **kwargs):
        super().__init__(index=None, **kwargs)
        self.embedder = embedder
        self.conn = conn

    def search(self, query, num_results=5):
        query_vector = self.embedder.encode(query)
        query_str = vec_to_str(query_vector)

        rows = self.conn.execute(
            """
            SELECT code, name, symbol, decimal_digits, countries
            FROM currencies
            ORDER BY embedding <=> %s::vector
            LIMIT %s
            """,
            (query_str, num_results)
        ).fetchall()

        return [
            {
                "code": r[0],
                "name": r[1],
                "symbol": r[2],
                "decimal_digits": r[3],
                "countries": r[4]
            }
            for r in rows
        ]
```

```python
from rag_helper import RAGBase

class RAGVector(RAGBase):

    def __init__(self, embedder, **kwargs):
        super().__init__(**kwargs)
        self.embedder = embedder

    def search(self, query, num_results=5):
        query_vector = self.embedder.encode(query)

        return self.index.search(
            query_vector,
            num_results=num_results
        )
```

```python
vector_assistant = RAGVector(
    index=vs_index,
    embedder=model,
    llm_client=openai_client
)

print(vector_assistant.rag(
    "What is the currency code and symbol for the Japanese Yen?"
))
```
